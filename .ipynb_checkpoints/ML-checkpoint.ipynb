# Data Cleaning

There are two issues to solve:

**The first one is to drop duplicates.**

**The second one is to fix unrealistic data.**

```python
df = df.drop_duplicates()
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].clip(lower=0)
```

## Skewness

```python
num_cols = df.select_dtypes(include='number')
num_cols.skew()
```

## Feature Engineering

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python

def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_countries)
```

```python
df['Grouped_country'].value_counts()
```

Country has 111 unique values in this dataset — one-hot encoding that directly would add 110+ mostly-empty columns, which hurts the model far more than it helps (this is called high cardinality).and we need to fix it so The fix: keep the top 10 most frequent countries as their own category, and bucket everything else into "Other". We keep the signal that matters and lose the noise that doesn't.

```python
df.head()
```

Train-Test-Split

```python
from sklearn.model_selection import train_test_split

skwewd_col         = ['Study_Hours']
other_numeric_cols = ["Age", "Avg_Daily_Usage_Hours", "Daily_Unlocks","Physical_Activity_Hours",
                      "Sleep_Hours_Per_Night"]
ordinal_col        = ['Stress_Level']
normal_col         = ["Gender", "Academic_Level", "Most_Used_Platform", "Purpose_Of_Use", "Grouped_country"]


feature_col = skwewd_col + other_numeric_cols + ordinal_col + normal_col
X = df[feature_col]
y = df['Mental_Health_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

## Pre Processing

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

* standard scaler is used to convert all the values on like one comparable scale
* ordinal encoder converts the categorical column on the basis on their order
* one hot encoder converts into separate 0/1 columns

* column transformer will manage all of these about mentioned thingy

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
```

```python
ordinal_encoder = OrdinalEncoder(
    categories=[['Low', 'Medium', 'High', 'Very High']]
)
```

```python
onehot_encoder = OneHotEncoder(
    handle_unknown='ignore'
)
```

```python
numerical_cols = skwewd_col + other_numeric_cols

scaler = StandardScaler()
print(numeric_cols)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', scaler, numeric_cols),
        ('ord', ordinal_encoder, ordinal_col),
        ('cat', onehot_encoder, normal_col)
    ]
)
```

```python
print(X_train.columns.tolist())
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
print(X_train.shape)
print(X_train_processed.shape)
```

```python
X_test_processed = preprocessor.transform(X_test)

print(X_test.shape)
print(X_test_processed.shape)
```

## Model Training

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_processed, y_train)
```

```python
y_pred = model.predict(X_test_processed)
```

```python
y_pred[:10]
```

```python
results = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

results.head(10)
```

we cant say much by just looking at these 10 first rows so we should calculate the error off alllll rows so 


## Model Evaluation

we will check threee metrics for the evalution
* MAE
* RMSE
* R²

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```
